<details><summary>Что такое защита проекта и чем она отличается от демонстрации</summary>

Демонстрация показывает, что система работает. Защита показывает, что ты **знаешь, насколько
она работает и почему**. Это разные жанры, и второй заметно труднее.

**Демонстрация звучит так.** «Мы собрали гибридный поиск с переранжированием, nDCG@10 = 0,64,
вот примеры выдачи, всё найдено.» Проверить нечего: неизвестно, что было бы без гибрида,
неизвестен разброс, неизвестно, лучше ли это BM25 из первой недели.

**Защита звучит так.** «База — BM25, MRR@10 = 0,27. Полная система даёт 0,35, это плюс двадцать
семь процентов относительных. Абляции: убрав переранжирование, теряем 0,009 при интервале,
накрывающем ноль; убрав плотную ступень — 0,032, интервал тоже накрывает ноль. То есть **ни один
компонент по отдельности не доказан** на ста пятидесяти запросах. Потолок объединения ретриверов
0,79 — то есть на пятой части запросов нужный фрагмент не нашёл никто; узкое место в отборе,
а не в переранжировании, и это следующий шаг.»

**Что различает эти два текста.** Второй **можно оспорить**: в нём есть база, интервалы,
потолки и названная слабость. Именно это делает его защитой. Текст, который нельзя оспорить,
потому что в нём нет проверяемых утверждений, — не защита, а презентация.

**Главное, что стоит понять до начала.** Отрицательный результат в защите — сила, а не слабость.
«Мы попробовали X, прирост оказался внутри интервала, и мы его не внедрили» показывает,
что ты умеешь останавливаться. Умение добавлять есть у всех; умение не добавлять — редкое
и дорогое.
</details>

<details><summary>Как читать чужую заголовочную таблицу: пять вопросов за минуту</summary>

Навык, который останется после курса дольше всего: смотреть на чужую таблицу результатов
и за минуту понимать, чему верить. Пять вопросов, в порядке убывания полезности.

**Первый: с чем сравнивают?** Если базы нет вовсе или это «random» — читать дальше незачем.
Если база есть, но не BM25 — спросить почему: BM25 бесплатен, и его отсутствие обычно означает,
что он выигрывал.

**Второй: одинаковы ли условия у строк?** Наша сегодняшняя таблица провалила бы этот вопрос:
у полной системы выдача двадцать пять, у остальных сто. В чужих работах это чаще всего разное
число кандидатов, разная глубина отсечки или разный препроцессинг.

**Третий: есть ли разброс?** Точечные оценки без интервалов означают, что автор либо не считал,
либо посчитал и не показал. Оба варианта одинаково информативны.

**Четвёртый: сколько запросов?** Число, которое обязано быть в подписи и которого там обычно
нет. Пятьдесят запросов и разница в две сотых — это ничего.

**Пятый: есть ли абляции?** Система из пяти компонентов без абляций — это утверждение, что все
пять нужны, не подкреплённое ничем. Часто оказывается, что работает один.

**Чего в этом списке нет.** Величины метрики. Она интересна последней, потому что зависит
от коллекции, разметки и постановки — то есть несравнима ни с чем за пределами самой таблицы.
Сравнивать надо **разницы внутри** таблицы, и то после четырёх вопросов выше.

**Практика, которая быстро ставит навык.** Возьми любую статью по поиску за последний год,
задай эти пять вопросов её главной таблице и посчитай, на сколько нашёлся ответ. Обычно
на два-три. Это не значит, что работа плоха; это значит, что читать её надо иначе, чем
она себя подаёт.
</details>

## Шаг 0 · Пины и preflight

<details><summary>Три способа провалить проект, которые повторяются каждый год</summary>

Не про технику, а про то, как расходуется время. Все три предсказуемы и все три лечатся
порядком работы.

**Первый: начать с самого интересного.** Агентный RAG, дообучение, своя архитектура —
всё это притягательно и всё это надо делать **после** того, как заработала база и собрана
разметка. Человек, потративший три недели на агентов, приходит на защиту без бейзлайна,
и всю его работу нечем оценить: не с чем сравнить.

Признак, что ты в этой ловушке: у тебя есть код нескольких моделей и нет ни одного числа,
посчитанного на своей разметке.

**Второй: откладывать разметку.** Она скучная, и её всегда можно отложить «до завтра».
Без неё нельзя измерить ничего, поэтому вся работа до неё делается вслепую: ты крутишь
параметры, глядя на три примера выдачи, и убеждаешь себя, что стало лучше. Разметка —
первое, что надо сделать после выбора корпуса, а не последнее.

Признак: ты подбирал параметры, но не можешь сказать, на чём мерил.

**Третий: строить в ширину, а не в глубину.** Соблазн добавить ещё одну ступень вместо того,
чтобы разобраться в имеющейся. Система из шести компонентов, ни один из которых не измерен,
слабее двухступенчатой, где обе ступени измерены и обоснованы. Каждая новая ступень
увеличивает число объяснений, которые придётся дать на защите.

Признак: в твоём конвейере больше ступеней, чем строк в таблице абляций.

**Общее лекарство.** Работать по кругу: минимальная система целиком → измерение → одно
улучшение → измерение. Каждый круг короткий, и после каждого у тебя есть **работающая
и измеренная** система. Если время кончится посреди круга, сдавать будет что.

**Обратная сторона.** Такой порядок кажется медленным первые две недели и оказывается
единственным, который доходит до финиша. Это тот же принцип, что и во всём курсе: сначала база,
потом улучшение, и каждое улучшение с интервалом.
</details>

| 1.2 | Четыре ступени | чанкование, лексика, плотность, слияние, переранжирование |
| 1.3 | Заголовочная таблица | то, с чего начинается защита — и что в ней уже сломано |

````python
raw = fetch_20newsgroups(subset="all", remove=("headers", "footers", "quotes"),
                         random_state=SEED)
TOKEN = re.compile(r"[a-z]{2,}")
SENT = re.compile(r"(?<=[.!?])\s+")
pool = []
for t in raw.data:
    t = " ".join(t.split())
    if len(TOKEN.findall(t.lower())) < 150:
        continue
    ss = [s for s in SENT.split(t) if 12 <= len(s.split()) <= 25]
    if ss:
        pool.append((t, ss))
random.Random(SEED).shuffle(pool)

N = min(N_DOCS, len(pool))
DOCS, QUERIES, GOLD_W = [], [], []
for i, (text, ss) in enumerate(pool[:N]):
    if i < N_QUERIES:
        s = max(ss, key=len)
        pos = text.index(s)
        d = (text[:pos] + text[pos + len(s):]).strip()
        DOCS.append(d)
        QUERIES.append(s.strip())
        GOLD_W.append(min(len(text[:pos].split()), max(0, len(d.split()) - 1)))
    else:
        DOCS.append(text)

CHUNKS, OWNER = [], []
GOLD = [set() for _ in range(N_QUERIES)]
for di, d in enumerate(DOCS):
    w = d.split()
    for a in range(0, max(1, len(w)), CHUNK_WORDS):
        piece = w[a:a + CHUNK_WORDS]
        if not piece:
            break
        if di < N_QUERIES and a <= GOLD_W[di] < a + len(piece):
            GOLD[di].add(len(CHUNKS))
        CHUNKS.append(" ".join(piece))
        OWNER.append(di)
        if a + CHUNK_WORDS >= len(w):
            break

print(f"документов: {len(DOCS)} · чанков: {len(CHUNKS)} "
      f"({len(CHUNKS) / len(DOCS):.1f} на документ)")
print(f"запросов: {len(QUERIES)} · золотых чанков на запрос: "
      f"{statistics.mean(len(g) for g in GOLD):.1f}")
RUN["n_chunks"] = len(CHUNKS)
````

**Что видно.** Корпус превратился в чанки, и их в несколько раз больше, чем документов.
Сравнивать надо не эти два числа, а **среднее число золотых чанков на запрос**: оно близко
к единице, потому что перекрытия нет, и позицию ответа покрывает ровно одно окно. Механизм
из недели 12: без перекрытия золотой чанк единственен, с перекрытием их было бы больше,
и засчитывать надо любой. Чего этот вывод НЕ показывает: качества нарезки — мы режем
по фиксированному числу слов и рвём предложения посередине. Что делать: для своего проекта
посмотреть глазами на десяток чанков. Если корпус структурирован — документация, вики, —
резать надо по заголовкам, и это единственное место, где стоит отступить от нашей конфигурации
без дополнительных замеров.

<details><summary>Чек-поинты проекта: зачем они и что сдавать на каждом</summary>

Проект разбит на чек-поинты не для отчётности, а потому что три типичных провала случаются
именно в тех местах, где стоят ворота.

**Первый чек-поинт: корпус выбран, постановка написана, двадцать запросов сформулированы.**
Обратная связь, без оценки. Провал, который он ловит: корпус, в котором нечего искать.
Если двадцать запросов не придумываются за вечер, дальше идти нельзя — весь проект повиснет
на выдуманных примерах. Лучше сменить корпус на первой неделе, чем на последней.

**Второй чек-поинт: разметка собрана и посчитан бейзлайн BM25.** Ворота жёсткие: нет числа
базы — дальше не оценивается. Провал, который он ловит: работа началась с нейросетей.
Человек три недели собирает эмбеддинги, потом считает BM25 и обнаруживает, что тот выигрывает.
Порядок обратный: сначала база, и она задаёт, есть ли вообще о чём говорить.

**Третий чек-поинт: нейронная ступень встроена, первая таблица приростов.** Провал, который
он ловит: система, которая не собирается целиком. Отдельно работающий индекс и отдельно
работающий реранкер — это не система, и склеить их в последнюю ночь не получится.

**Четвёртый — защита.** Всё, что мы делали сегодня.

**Что общего у первых трёх.** Каждый требует **числа**, а не описания. «Корпус выбран,
двадцать запросов есть» проверяется списком; «база посчитана» — числом; «ступень встроена» —
таблицей. Ни один не проверяется словами «работаю над этим».

**И про сроки.** Разметка занимает больше времени, чем кажется, и меньше, чем страшно.
Сто запросов с бинарной релевантностью по топ-10 — это вечер, если инструмент простой:
таблица, где ты ставишь галочки. Написать этот инструмент — полчаса, и он окупается
пятикратно, потому что размечать придётся не один раз.
</details>

<details><summary>Как выбрать корпус для проекта — и три ошибки выбора</summary>

Задание разрешает любой легальный корпус от пяти тысяч документов. Выбор влияет на проект
сильнее, чем любое техническое решение, и ошибиться легко.

**Ошибка первая: корпус без естественных запросов.** Дамп википедии красив и бесполезен:
непонятно, что там искать. Хороший признак корпуса — существование людей, которые уже сейчас
что-то в нём ищут: документация продукта, архив форума, объявления, научные абстракты своей
области. Если ты сам не можешь придумать двадцать запросов за десять минут, корпус плохой.

**Ошибка вторая: слишком однородный корпус.** Пять тысяч почти одинаковых карточек товара
дадут поиск, где всё похоже на всё, и различить методы будет нечем: и BM25, и плотный
покажут одинаковые числа. Полезен корпус с разнообразием тем — тогда разные методы промахиваются
по-разному, и появляется, что сравнивать.

**Ошибка третья: корпус, который нельзя разметить.** Если для суждения о релевантности нужен
эксперт — юридические тексты, медицина, узкая инженерия, — ты не сможешь собрать разметку сам,
и весь проект повиснет на псевдозапросах. Это допустимо, но названо должно быть сразу.

**Что делает корпус хорошим.** Ты в нём разбираешься настолько, чтобы разметить сто запросов
за вечер. Документы разной длины и тематики. Существуют запросы, где лексика точно не сработает
(синонимия, перефразировка), и запросы, где точно сработает только она (артикулы, имена,
номера). Тогда гибрид будет иметь смысл, и его можно будет защитить.

**И про лицензию.** Правило 11.1 курса: только явно открытые лицензии. Дамп форума без
разрешения — не проект, а проблема, и на защите это первый вопрос.
</details>

<details><summary>Что должно быть в проекте помимо чисел: инженерная часть</summary>

Защита строится на числах, но проект — это работающая система, и у неё есть требования,
которых метрики не покрывают.

**Воспроизводимость одной командой.** Клонировать репозиторий, выполнить одну команду, получить
те же числа. Если для этого нужно вручную скачать датасет, положить его в правильную папку
и раскомментировать строку — проект не воспроизводим. Скрипт, который делает всё сам,
пишется за полчаса и снимает половину вопросов.

**Разделение индексации и поиска.** Индексация — долгая офлайн-операция, поиск — быстрая
онлайн. Если они в одном скрипте и корпус переиндексируется на каждый запрос, это не система,
а демонстрация. Артефакт индекса должен сохраняться на диск и загружаться.

**Обработка пустого результата.** Что система отвечает, когда ничего не найдено? Пустая выдача,
сообщение, топ-k мусора? Молчаливый возврат мусора — самый частый и самый плохой вариант.

**Логирование запроса и того, что нашлось.** Без этого невозможно ни разобрать жалобу,
ни собрать данные для следующей итерации. Три строки кода, которые окупаются на первой же
непонятной выдаче.

**Конфигурация в одном месте.** Все ручки — размер чанка, `alpha`, глубина — в одном файле
или блоке, а не разбросаны по коду. Мы так и сделали в шаге 1, и это не стилистическое
предпочтение: конфигурация, которую нельзя увидеть целиком, не может быть защищена целиком.

**Чего не требуется.** Веб-интерфейса, докера, CI. Проект оценивается по системе и её защите,
а не по обвязке. Время, потраченное на красивый фронтенд, лучше потратить на пятьдесят
дополнительных размеченных запросов — они изменят выводы, а фронтенд нет.
</details>

⚠️ Ловушка A · **Псевдозапросы — не пользовательские запросы, и это ограничение всего курса.**
Мы использовали их семь занятий подряд, потому что они дают воспроизводимую разметку бесплатно.
Плата: они лексически щедры к BM25, длиннее реальных в разы и написаны автором документа,
а не человеком, который ищет. Для **сравнения конфигураций** этого достаточно — смещение
одинаково давит на все. Для **абсолютных чисел** — нет, и ни одно значение из этого ноутбука
не следует называть вслух без этой оговорки.

«качество по неделям курса» велик и совершенно незаконен: это семь разных экспериментов,
а не одна кривая.

### Шаг 1.2 · Четыре ступени, собранные вместе

Лексическая, плотная, слияние, переранжирование. Каждая — с конкретной недели, и код почти
дословно оттуда.

````python
st = SentenceTransformer(BI_MODEL)
t0 = time.perf_counter()
CEMB = st.encode(CHUNKS, batch_size=256, normalize_embeddings=True,
                 show_progress_bar=False).astype("float32")
QEMB = st.encode(QUERIES, normalize_embeddings=True, show_progress_bar=False).astype("float32")
t_index = time.perf_counter() - t0
SIM = QEMB @ CEMB.T
DENSE = [[int(x) for x in np.argsort(-SIM[i])[:TOP_K]] for i in range(N_QUERIES)]
DENSE_S = [[float(SIM[i][d]) for d in DENSE[i]] for i in range(N_QUERIES)]

post, dl = defaultdict(dict), {}
for i, c in enumerate(CHUNKS):
    tk = TOKEN.findall(c.lower())
    dl[i] = len(tk)
    for t, tf in Counter(tk).items():
        post[t][i] = tf
NC, AVGDL = len(CHUNKS), sum(dl.values()) / len(CHUNKS)

def bm25(query, k1=1.5, b=0.75):
    sc = defaultdict(float)
    for t in TOKEN.findall(query.lower()):
        p = post.get(t)
        if not p:
            continue
        idf = math.log((NC - len(p) + 0.5) / (len(p) + 0.5) + 1)
        for d, tf in p.items():
            sc[d] += idf * (tf * (k1 + 1)) / (tf + k1 * (1 - b + b * dl[d] / AVGDL))
    o = sorted(sc, key=lambda d: (-sc[d], d))[:TOP_K]
    return o, [sc[d] for d in o]

LEX, LEX_S = zip(*(bm25(q) for q in QUERIES))
LEX, LEX_S = list(LEX), list(LEX_S)

def zscore(d):
    if not d:
        return {}
    v = np.fromiter(d.values(), float)
    mu, sd = v.mean(), v.std()
    return {k: ((x - mu) / sd if sd > 0 else 0.0) for k, x in d.items()}

def fuse(i, alpha=ALPHA):
    a = zscore(dict(zip(LEX[i], LEX_S[i])))
    b = zscore(dict(zip(DENSE[i], DENSE_S[i])))
    cand = set(a) | set(b)
    sc = {d: alpha * a.get(d, -4.0) + (1 - alpha) * b.get(d, -4.0) for d in cand}
    return sorted(sc, key=lambda d: (-sc[d], d))[:TOP_K]

FUSED = [fuse(i) for i in range(N_QUERIES)]

ce = CrossEncoder(CE_MODEL)
def rerank(base, depth=RERANK_DEPTH):
    out, t0 = [], time.perf_counter()
    for i, q in enumerate(QUERIES):
        cand = base[i][:depth]
        s = ce.predict([(q, CHUNKS[d]) for d in cand], batch_size=32, show_progress_bar=False)
        out.append([int(cand[j]) for j in np.argsort(-s)])
    return out, time.perf_counter() - t0

FULL, t_rerank = rerank(FUSED)
print(f"индексация: {t_index:.0f} c · переранжирование {N_QUERIES} запросов: {t_rerank:.0f} c "
      f"({t_rerank / N_QUERIES * 1000:.0f} мс/запрос)")
print(f"длина выдачи: лексика {len(LEX[0])} · плотный {len(DENSE[0])} · "
      f"слияние {len(FUSED[0])} · ПОЛНАЯ {len(FULL[0])}")
RUN["index_sec"], RUN["rerank_ms"] = t_index, t_rerank / N_QUERIES * 1000
````

**Что видно.** Последняя строка — самая важная в этой ячейке, и на неё обычно не смотрят.
Сравнивать надо **длину выдачи полной системы с длинами остальных**: у трёх ступеней по сто
результатов, у полной — двадцать пять, потому что переранжирование работает на глубине
двадцать пять и физически не может вернуть больше. Механизм тривиален и потому опасен: любая
метрика с отсечкой глубже двадцати пяти у полной системы **обязана** проиграть, и не потому,
что система хуже. Чего этот вывод НЕ показывает: во что это выльется — увидим в заголовочной
таблице через ячейку. Что делать: запомнить это число. Мы сейчас построим таблицу, в которой
оно испортит одну колонку, и специально не станем её чинить сразу.

<details><summary>Что померять для защиты помимо качества: цена системы</summary>

Мы измеряли качество семь занятий. Защита требует ещё трёх чисел, которые обычно забывают.

**Время индексации.** Сколько занимает построение системы с нуля. У нас — десятки секунд
на две с половиной тысячи документов, то есть порядка десяти миллисекунд на документ.
На корпусе в миллион это часы, и это уже операционное ограничение: переиндексация после смены
модели становится событием, которое планируют.

**Задержка запроса по ступеням.** Не общая, а разложенная: кодирование запроса, поиск,
слияние, переранжирование. Тогда видно, что оптимизировать. У нас переранжирование доминирует
с огромным отрывом, и это типично.

**Память.** Векторы, индекс, постинги, исходные тексты. Неделя 10 показала, что именно память,
а не время, задаёт границу применимости. Для защиты достаточно четырёх чисел и суммы.

**Почему это часть защиты, а не приложение.** Потому что все решения курса были компромиссами
между качеством и ценой, и без цены они не решения. «Мы выбрали глубину двадцать пять» —
неполное утверждение; полное: «глубина двадцать пять, потому что сто дороже вчетверо
и не отличима по качеству».

**Как мерить, чтобы не соврать.** Прогрев обязателен (неделя 10 показала, что без него первая
строка вдвое хуже). Число потоков — записать. Батч — записать. И мерить **и** пропускную
способность, **и** задержку одиночного запроса: в проде запросы приходят по одному, а батч
льстит.

**Одно число, которое стоит всех.** Стоимость одного запроса в деньгах. Секунды GPU
и мегабайты памяти переводятся в рубли по прайсу облака, и разговор о том, нужна ли третья
ступень, сразу становится предметным.
</details>

<details><summary>Почему обрезка выдачи — самая живучая ошибка курса</summary>

Она встретилась трижды: на неделе 9, на неделе 12 и сегодня в собственной заголовочной таблице.
Стоит понять, почему её так трудно заметить.

**Она не выглядит ошибкой.** Таблица правильная: столбцы подписаны, числа посчитаны верно,
формула метрики одна и та же. Ничто в таблице не намекает, что у одной строки выдача короче.

**Она возникает естественно.** Переранжирование по определению работает на глубине `d`
и возвращает `d` результатов. Никто не решал «обрежем выдачу» — так получилось из устройства
ступени.

**Она бьёт в нужную сторону.** Обрезка портит метрики с глубокой отсечкой (R@100) и не трогает
мелкие (MRR@10, R@10). А в отчёты обычно кладут именно мелкие — и там всё выглядит прекрасно.
Ошибка проявляется только в той колонке, на которую смотрят меньше всего.

**Как ловить.** Правило простое и механическое: **в сравнительную таблицу добавляй колонку
с длиной выдачи**. Мы так и сделали, и обрезка стала видна сразу. Если длины различаются,
любая метрика с отсечкой глубже минимальной из них — недействительна.

**Как чинить.** Два честных варианта. Первый: мерить всё на глубине, не превышающей минимальную
выдачу. Второй: доукомплектовывать короткую выдачу хвостом входного ранжирования — то есть
после переранжирования двадцати пяти приписывать остальные семьдесят пять в исходном порядке.
Второй вариант ближе к тому, как система работает в проде, и потому предпочтительнее.

**Чего делать нельзя.** Оставлять колонку и приписывать сноску «у полной системы выдача
короче». Сноску не читают, а число сравнивают.
</details>

### Шаг 1.3 · Заголовочная таблица

То, с чего начинается любая защита. И то, в чём почти всегда что-нибудь сломано.

````python
def rr_at(order, gold, k=10):
    for i, d in enumerate(order[:k], 1):
        if d in gold:
            return 1 / i
    return 0.0

def recall_at(order, gold, k):
    return 1.0 if gold & set(order[:k]) else 0.0

SYSTEMS = {"BM25 (неделя 3)": LEX, "плотный (неделя 7)": DENSE,
           "слияние (неделя 9)": FUSED, "ПОЛНАЯ СИСТЕМА": FULL}

def mrr(orders):
    return statistics.mean(rr_at(orders[i], GOLD[i]) for i in range(N_QUERIES))

def rec(orders, k):
    return statistics.mean(recall_at(orders[i], GOLD[i], k) for i in range(N_QUERIES))

print(f"{'система':>22} {'MRR@10':>9} {'R@10':>8} {'R@100':>8} {'выдача':>8}")
head = {}
for name, o in SYSTEMS.items():
    head[name] = {"mrr": mrr(o), "r10": rec(o, 10), "r100": rec(o, 100), "len": len(o[0])}
    h = head[name]
    print(f"{name:>22} {h['mrr']:>9.4f} {h['r10']:>8.3f} {h['r100']:>8.3f} {h['len']:>8}")
base_mrr = head["BM25 (неделя 3)"]["mrr"]
full_mrr = head["ПОЛНАЯ СИСТЕМА"]["mrr"]
print(f"\nполная система против базы: {(full_mrr / base_mrr - 1) * 100:+.1f}% относительных "
      f"по MRR@10")
RUN["headline"] = head
````

**Что видно.** Заголовок выглядит убедительно: полная система обыгрывает BM25 на четверть
с лишним по MRR@10. И тут же, в той же таблице, лежит подлог — **колонка R@100 у полной системы
ниже, чем у слияния**. Сравнивать надо не системы по этой колонке, а **колонку R@100 с колонкой
длины выдачи**: у полной системы двадцать пять результатов, и её «R@100» — это на самом деле
R@25. Механизм разобран ячейкой выше, и мы намеренно оставили таблицу как есть, потому что
именно так она обычно и попадает в защиту. Чего этот вывод НЕ показывает: какая часть прироста
MRR настоящая — это следующая часть. Что делать: вычеркнуть колонку R@100 либо померять
все системы на глубине двадцать пять. Первое честнее: метрика, которую нельзя посчитать
для всех сравниваемых, в сравнительной таблице не место.

<details><summary>Целое доказано, части — нет: что это значит и как об этом говорить</summary>

Наш результат устроен неудобно: полная система против базы отличима, а каждая отдельная
абляция — нет. Это не противоречие и не ошибка, и стоит понимать, почему.

**Почему так получается арифметически.** Эффект целого — это сумма вкладов трёх компонентов
плюс их взаимодействия. Сумма может быть заметной, а каждое слагаемое по отдельности —
меньше разрешения замера. Разрешение при этом одинаково для всех сравнений: оно определяется
разбросом и числом запросов, а не величиной эффекта.

**Что отсюда следует практически.** Утверждение «наша система лучше базы» **доказано**
и может идти в защиту без оговорок. Утверждение «переранжирование даёт вклад» — **не доказано**,
и должно идти с интервалом и словами «на нашей выборке различить не удалось».

**Чего делать нельзя.** Первое: выбрасывать компоненты по результатам абляций. Три
недоказанных вклада, сложенные вместе, дают доказанный эффект — выбросив все три, ты потеряешь
его целиком. Второе: заявлять, что «каждый компонент вносит вклад», ссылаясь на отличимость
целого. Это подмена, и на защите её ловят первым вопросом.

**Как говорить честно.** «Система в целом обыгрывает базу, интервал такой-то. Разложить эффект
по компонентам на этой выборке не удалось: все три абляции внутри шума. Чтобы разложить,
нужно примерно вчетверо больше запросов, и мы оценили это заранее.»

**Почему это сильная позиция, а не слабая.** Потому что она предъявляет ровно то, что известно,
и ровно то, что неизвестно, с ценой перехода из второго в первое. Защита, где всё доказано,
вызывает вопрос «а вы точно проверяли?»; защита, где чётко разделены доказанное и нет, вопросов
не вызывает.

**И одно наблюдение про курс.** Такая картина — целое отличимо, части нет — типична для
инженерных систем и почти никогда не встречается в статьях. Причина простая: статья, где
компоненты не доказаны, не публикуется, поэтому в литературе виден только тот срез, где данных
хватило. Это ещё одна причина не сравнивать свои числа с опубликованными.
</details>

<details><summary>Что делать, когда защита показала «ничего не доказано»</summary>

Наш случай, и он гораздо более частый, чем кажется по статьям. Четыре честных пути,
и ни один из них не «переделать эксперимент, пока не получится».

**Первый: набрать разметку.** Прямой и единственный путь к разрешению. Мы посчитали:
для эффекта в три сотых при нашем разбросе нужно порядка шестисот запросов вместо ста
пятидесяти. Псевдозапросы генерируются бесплатно, так что это часы счёта, а не недели.
Для реальной разметки — недели, и тогда вопрос переходит в плоскость «стоит ли эффект
этих недель».

**Второй: уменьшить разброс.** Метрика с меньшей дисперсией даёт то же разрешение на меньшей
выборке. MRR по одному релевантному документу — худшее, что можно выбрать; nDCG по
градуированной разметке шумит заметно меньше. Плата — нужна разметка с градациями.

**Третий: решать по цене.** Если интервал накрывает ноль, а компонент стоит сто миллисекунд
на запрос, решение принимается по цене: не доказано и дорого — не внедряем. Это **полноценный
инженерный вывод**, а не отговорка, и в защите он звучит сильно.

**Четвёртый: сменить вопрос.** Мы не смогли доказать вклад компонентов, но уверенно нашли
узкое место лестницей потолков. Исходный вопрос не разрешился имеющимися данными, соседний —
разрешился, и ответ на него полезнее.

**Чего делать нельзя.** Добирать запросы, пока интервал не перестанет накрывать ноль,
и останавливаться в этот момент. Это optional stopping, и оно гарантированно даёт
ложноположительный результат при достаточном терпении. Размер выборки фиксируется до замера.

**Формулировка для защиты.** «Наш замер не обладает разрешением для эффектов такого размера;
чтобы его получить, нужно вчетверо больше запросов. Мы приняли решение по цене и назвали
условие, при котором его пересмотрим.» Это сильнее, чем любое «значимо».
</details>

## Часть 2 · Абляции — 30 мин

| шаг | вопрос | чем отвечаем |
|---|---|---|
| 2.1 | Что даёт каждый компонент? | убираем по одному и меряем разницу |
| 2.2 | Отличимо ли это от шума? | парный тест по методике недели 4 |

Абляция — единственный способ узнать, что именно работает. Заголовочная таблица сравнивает
**разные системы**; абляция сравнивает систему **с собой минус один компонент**, и только
это отвечает на вопрос «зачем он здесь».

````python
ABL = {}
ABL["без переранжирования"] = FUSED
ABL["без слияния (плотный + переранжирование)"] = rerank(DENSE)[0]
ABL["без плотного (BM25 + переранжирование)"] = rerank(LEX)[0]

def per_query(orders):
    return [rr_at(orders[i], GOLD[i]) for i in range(N_QUERIES)]

def paired(a, b):
    d = [x - y for x, y in zip(a, b)]
    m = statistics.mean(d)
    se = statistics.stdev(d) / math.sqrt(len(d))
    return m, m - 2.0 * se, m + 2.0 * se

full_pq = per_query(FULL)
print(f"{'конфигурация':>42} {'MRR@10':>9} {'Δ к полной':>12} {'ДИ95':>24} вердикт")
print(f"{'ПОЛНАЯ СИСТЕМА':>42} {statistics.mean(full_pq):>9.4f} {'--':>12} {'--':>24}")
ablations = {}
for name, orders in ABL.items():
    pq = per_query(orders)
    m, lo, hi = paired(pq, full_pq)
    verdict = "РАЗЛИЧИМО" if lo > 0 or hi < 0 else "не отличимо"
    ablations[name] = {"mrr": statistics.mean(pq), "delta": m, "ci": [lo, hi],
                       "distinguishable": lo > 0 or hi < 0}
    print(f"{name:>42} {statistics.mean(pq):>9.4f} {m:>+12.4f} "
          f"[{lo:+.4f}; {hi:+.4f}] {verdict}")
RUN["ablations"] = ablations
````

**Что видно.** Убрав любой компонент, мы теряем — и **ни одна потеря не отличима от нуля**.
Сравнивать надо не величины Δ между собой, а **каждый интервал с нулём**: все три накрывают
его. То есть заголовочные «плюс двадцать семь процентов» получены системой, в которой ни один
компонент не доказан по отдельности. Механизм известен с недели 4: сто пятьдесят запросов
при разбросе MRR по одному релевантному документу дают разрешение порядка нескольких сотых,
а наши эффекты — того же порядка. Чего эта таблица НЕ показывает: что компоненты бесполезны.
Отсутствие доказательства — не доказательство отсутствия, и это разные утверждения.
Что делать: не выбрасывать компоненты по результату этой таблицы. Она говорит, что **замер
не обладает разрешением**, а не что компоненты не нужны.

<details><summary>Как выбирать, что исключать — и почему трёх абляций мало</summary>

Мы убрали три компонента. Компонентов в системе больше, и выбор того, что исключать, —
отдельное решение.

**Полный список ручек нашей системы.** Размер чанка, перекрытие, модель эмбеддингов, `alpha`
слияния, нормировка скоров, заполнитель отсутствующих, глубина переранжирования, модель
кросс-энкодера, глубина отбора `TOP_K`. Девять, и это ещё скромно.

**Что исключать в первую очередь.** То, что **дорого**. Переранжирование стоит сотню
миллисекунд на запрос — если оно не доказано, это самая дорогая недоказанная строка
в бюджете. Плотная ступень требует GPU для индексации и памяти под векторы. Слияние
бесплатно, и его абляция — приятная, но малополезная.

**Что исключать во вторую очередь.** То, где ты **меньше всего уверен**. Если `alpha`
подбиралась на отложенной половине и разница между 0,2 и 0,5 была внутри шума, стоит проверить
крайние значения: вдруг лексика вообще не нужна.

**Чего абляции не покрывают.** Взаимодействий. Убрав A, ты меряешь эффект A **при наличии
всех остальных**. Возможно, A бесполезен с B, но необходим без B. Полная картина требует
перебора подмножеств, то есть двойки в степени числа компонентов, и на практике не делается
никогда.

**Практический компромисс.** Одна абляция на каждый **дорогой** компонент плюс одна на самый
сомнительный. Четыре-пять строк в таблице. Больше — и защита превращается в отчёт
об экспериментах, а не в обоснование решения.
</details>

⚠️ Ловушка C · **«Не отличимо» — не «не нужно».** Соблазн очевиден: раз убирание
переранжирования не ухудшает статистически, выбросим его и сэкономим сто миллисекунд.
Это ошибка того же рода, что «p выше порога значит эффекта нет». Правильный вывод: **на нашей
выборке решение о переранжировании принять нельзя**, и если оно дорого, надо либо набрать
запросов, либо решать по цене, а не по метрике.

⚠️ Ловушка A · **Сумма абляций не равна целому.** Убрав переранжирование, теряем 0,009;
убрав плотную ступень — 0,032. Из этого **не следует**, что вместе они дают 0,041: компоненты
взаимодействуют. Плотная ступень поставляет кандидатов переранжированию, и без неё оно работает
с другим входом. Честная абляция каждого компонента считается отдельно и не складывается.

````python
fig, ax = plt.subplots(figsize=(9, 3))
names = list(ablations)
deltas = [ablations[n]["delta"] for n in names]
errs = [[ablations[n]["delta"] - ablations[n]["ci"][0] for n in names],
        [ablations[n]["ci"][1] - ablations[n]["delta"] for n in names]]
colors = ["#B4521F" if ablations[n]["distinguishable"] else "#8A8A8A" for n in names]
ax.barh(range(len(names)), deltas, xerr=errs, color=colors, alpha=.75,
        error_kw={"ecolor": "#111", "capsize": 4})
ax.axvline(0, color="#111", lw=1.5)
ax.set_yticks(range(len(names)))
ax.set_yticklabels([n.split(" (")[0] for n in names], fontsize=9)
ax.set_xlabel("изменение MRR@10 при удалении компонента (отрицательное = стало хуже)")
plt.title("Абляции: все интервалы пересекают ноль")
plt.tight_layout(); plt.show()
````

**Что видно.** Все три планки уходят влево — то есть без компонента хуже, — и все три усика
пересекают вертикальную линию нуля. Сравнивать надо не длины планок, а **пересечение усиков
с нулём**: пока оно есть, направление эффекта не установлено, каким бы убедительным
ни выглядел столбик. Ожидаемая картина по механизму из разбора выше. Чего график
НЕ показывает: сколько запросов понадобилось бы. Прикинуть можно по формуле недели 4:
ширина интервала падает как корень из числа запросов, значит для эффекта в 0,03 при нашем
разбросе нужно порядка шестисот запросов вместо ста пятидесяти. Что делать: включить эту
оценку в защиту. «Нам не хватило разрешения, и вот сколько его стоило бы» — сильное
утверждение, а «компоненты работают» без интервалов — слабое.

## Часть 3 · Лестница потолков — 25 мин

| шаг | вопрос | чем отвечаем |
|---|---|---|
| 3.1 | Сколько теряет каждая ступень? | **замер без модели**: потолок на входе каждой ступени |
| 3.2 | Где узкое место? | сравниваем потери между ступенями |

Абляции сказали, что мы не можем различить компоненты. Потолки скажут, **где искать
следующий шаг**, и для этого не нужна ни статистика, ни модели.

````python
ceil_union = statistics.mean(
    1.0 if GOLD[i] & (set(LEX[i][:TOP_K]) | set(DENSE[i][:TOP_K])) else 0.0
    for i in range(N_QUERIES))
ceil_fused = rec(FUSED, TOP_K)
ceil_into_rerank = rec(FUSED, RERANK_DEPTH)
achieved_r10 = rec(FULL, 10)
achieved_mrr = mrr(FULL)

ladder = [
    ("объединение лексики и плотного, глубина 100", ceil_union),
    ("после слияния, глубина 100", ceil_fused),
    ("дошло до переранжирования (глубина 25)", ceil_into_rerank),
    ("итоговая полнота в топ-10", achieved_r10),
]
print(f"{'ступень':>46} {'потолок':>9} {'потеря':>9}")
prev = 1.0
for name, v in ladder:
    print(f"{name:>46} {v:>9.3f} {v - prev:>+9.3f}")
    prev = v
print(f"\nитоговый MRR@10: {achieved_mrr:.4f}")
losses = {ladder[i][0]: ladder[i][1] - (ladder[i - 1][1] if i else 1.0)
          for i in range(len(ladder))}
worst = min(losses, key=losses.get)
print(f"наибольшая потеря на ступени: «{worst}» ({losses[worst]:+.3f})")
RUN["ladder"] = {n: v for n, v in ladder}
RUN["worst_stage"] = worst
````

**Что видно.** Лестница показывает то, чего не показала ни заголовочная таблица, ни абляции:
**где именно теряется полнота**. Сравнивать надо не абсолютные значения ступеней, а **колонку
потерь**, и самая большая — на самой первой строке. Это значит, что узкое место не в глубине
переранжирования и не в слиянии, а в **отборе как таковом**: для заметной доли запросов ни
лексический, ни плотный поиск не нашли нужный чанк в первой сотне вовсе. Никакая последующая
ступень этого не исправит. Вторая по величине потеря — обрезка выдачи со ста до двадцати пяти
перед переранжированием, и она вдвое меньше первой. Чего эта лестница НЕ показывает: сколько
из выброшенного было полезным — она считает только золотые чанки. Что делать: следующий шаг
теперь определён числом, а не вкусом, и он **не тот**, который напрашивался. Не «углубить
переранжирование», а «улучшить отбор»: другая модель эмбеддингов, дообучение на своём домене
или третий ретривер.

<details><summary>Почему узкое место оказалось не там, где ожидалось</summary>

Перед прогоном напрашивался ответ «обрезка перед переранжированием»: она видна, она груба,
она выбрасывает три четверти кандидатов. Лестница показала другое, и разбор этого расхождения
полезнее самого результата.

**Что показала лестница.** Самая большая потеря — на первой ступени: доля запросов, где нужный
чанк не нашёл **никто**, ни лексика, ни плотный поиск, ни в первой сотне. Обрезка перед
переранжированием стоит примерно вдвое меньше.

**Почему интуиция промахнулась.** Обрезка выглядит грубее, потому что видно, **сколько**
выбрасывается: семьдесят пять из ста. Но выбрасываются в основном плохие кандидаты — на местах
с двадцать шестого по сотое нужный чанк оказывается редко. А первая ступень теряет тихо:
ничего не выбрасывает, просто не находит, и в логах это никак не проявляется.

**Общее правило, которое отсюда следует.** Заметность потери и её величина не связаны. Ступень,
которая явно что-то отбрасывает, кажется главной подозреваемой; ступень, которая просто
не справляется, невидима. Поэтому потолки и считают — они меряют, а не оценивают на глаз.

**Что делать с таким диагнозом.** Улучшать отбор дороже, чем крутить глубину: нужна другая
модель эмбеддингов, дообучение на домене или третий ретривер. Зато потолок над этим действием
самый высокий из доступных. Правильная формулировка для защиты: «дешёвое действие даёт мало,
дорогое — много; мы выбрали такое-то и вот почему».

**И одна оговорка.** Наш первый шаг лестницы стартует от единицы, то есть измеряет полноту
относительно **идеала**, а не относительно предыдущей ступени. Это делает его завышенным
по сравнению с остальными по построению. Честнее было бы сравнивать потери, нормированные
на то, что было доступно на входе, — и это упражнение стоит сделать самому.
</details>

<details><summary>Лестница потолков как метод: почему она сильнее абляций</summary>

Абляции сказали «ничего не доказано». Лестница за одну ячейку показала, где узкое место.
Разница в природе этих двух инструментов, и её стоит понимать.

**Абляция — статистический инструмент.** Она сравнивает две системы и требует достаточной
выборки, чтобы отличить разницу от шума. При маленькой выборке она молчит — и это её честное
поведение, а не недостаток.

**Потолок — детерминированный.** Он не сравнивает системы, а считает, **что физически доступно**
следующей ступени. Никакого шума: либо нужный документ дошёл до ступени, либо нет. Дисперсия
есть только от выборки запросов, и она гораздо меньше, потому что величина бинарна
и не взвешена рангом.

**Отсюда практическое следствие.** При малой разметке потолки информативнее абляций. Полсотни
запросов уже дают осмысленную лестницу, а для различения эффекта в три сотых нужно шестьсот.
Если у тебя пятьдесят размеченных запросов — считай потолки и не пытайся доказывать эффекты.

**Что потолок не заменяет.** Он говорит, **где** предел, и молчит о том, стоит ли его двигать.
Узкое место может быть дорогим в починке и дешёвым в терпении. Решение остаётся инженерным,
и потолок — только вход в него.

**Форма, в которой это стоит запомнить.** Каждая ступень конвейера имеет вход и выход.
Потолок — это доля запросов, для которых правильный ответ **присутствовал на входе**. Разность
потолков соседних ступеней — потеря этой ступени. Сумма потерь плюс достигнутое равно единице.
Три строки кода на любой конвейер, от поиска до классификации.
</details>

⚠️ Ловушка B · **Потолок считается по золотым чанкам, а не по полезности.** Мы меряем, дошёл ли
до ступени тот единственный чанк, где лежал ответ. В реальной системе полезных чанков несколько,
и обрезка выбрасывает часть из них — потери на самом деле больше, чем показывает лестница.
Это смещение в сторону оптимизма, и его надо называть.

```python
losses_ = {}
prev_ = 1.0
for name_, v_ in ladder:
    losses_[name_] = v_ - prev_
    prev_ = v_
my_worst = min(losses_, key=losses_.get)
idx_ = [n for n, _ in ladder].index(my_worst)
headroom = (ladder[idx_ - 1][1] if idx_ else 1.0) - achieved_r10
```

Действие обязано следовать из диагноза, и диагноз здесь однозначен: самая большая потеря —
на **первой** строке лестницы, в объединении ретриверов. Значит действие — **улучшать отбор**:
другая модель эмбеддингов, дообучение на своём домене или третий ретривер. Цена известна —
переиндексация корпуса, у нас порядка минуты на две с половиной тысячи документов, на проде
считается заранее. Выигрыш ограничен сверху `headroom` — тем, что вообще доступно
следующей ступени.

Почему **не** «увеличить глубину переранжирования», хотя это первое, что напрашивается:
потеря на обрезке перед переранжированием вдвое меньше потери отбора, и неделя 7 показала,
что углубление добавляет судье конкурентов быстрее, чем правильных ответов. Правильная
формулировка следующего шага: «заменить эмбеддер, померить потолок объединения, и если он
не вырос — признать, что домен требует дообучения, а не другой готовой модели».

определяется иначе. Строго говоря, это не абляция, а **другая система**, и ставить её в один
столбец с остальными некорректно ровно по той же причине, по которой некорректна колонка R@100
в заголовочной таблице.

**Почему это кульминация занятия.** Самое убедительное число ноутбука — артефакт постановки,
и оно возникло не из подтасовки, а из честного выполнения задания. Ровно так выглядит
большинство впечатляющих результатов в чужих работах: сравнили не то с не тем, посчитали верно,

как главный вывод.

Правильно было бы либо сравнивать обе по полноте **документа**, либо признать, что чанкование —

<details><summary>Задание 3 · как выглядит хорошая защита</summary>

Признаки, по которым проверяющий отличит сильную защиту от слабой, — все шесть видны за минуту,

## Дамп прогона

Правило 10.5: занятие не считается прогнанным, пока его числа не лежат в файле рядом
с конфигурацией рантайма. Ячейка ниже собирает все численные результаты ноутбука —
от сида до финальных метрик — и кладёт их в `runs/project-search.json`. Это и есть
доказательство прогона: разбор сверяется с файлом, а не с памятью автора.

````python
# Итог прогона (правило 10.5): все числовые результаты + конфигурация рантайма +
# журнал печатей уезжают ОДНИМ архивом. Скачай его по ссылке ниже — и всё.
import json as _json, os as _os, sys as _sys, platform as _pl, shutil as _shutil, base64 as _b64

_runtime = {"python": _sys.version.split()[0], "platform": _pl.platform()}
_torch = _sys.modules.get("torch")   # НЕ импортируем сами: рамка 7.4 — сид и пин
if _torch is not None:                # обязателен только там, где ноутбук torch ИСПОЛЬЗУЕТ
    _runtime["torch"] = _torch.__version__
    _runtime["gpu"] = _torch.cuda.get_device_name(0) if _torch.cuda.is_available() else None
else:
    import subprocess as _sp
    try:
        _q = _sp.run(["nvidia-smi", "--query-gpu=name", "--format=csv,noheader"],
                     capture_output=True, text=True, timeout=5)
        _runtime["gpu"] = (_q.stdout.strip().splitlines() or [None])[0] if _q.returncode == 0 else None
    except Exception:
        _runtime["gpu"] = None

def _plain(v):
    try:
        import numpy as _np
        if isinstance(v, _np.integer): return int(v)
        if isinstance(v, _np.floating): return float(v)
    except Exception:
        pass
    return v

def _num(v):
    return isinstance(v, (int, float)) and not isinstance(v, bool)

def _tree(v, depth=0):
    """Числовое содержимое величины: само число, либо список/словарь чисел, в том числе
    ВЛОЖЕННЫЙ. Нечисловые ветки отбрасываются, пустое — тоже (None).

    Рекурсия здесь не украшение. Прежний сборщик брал словарь, только если ВСЕ его
    значения — числа, и поэтому целиком терял RUN: там лежит RUN["task1"] = {...},
    один вложенный словарь на весь ноутбук. А RUN — единственное место, где метрика
    записана в тех же единицах, в каких её печатает проза (RUN["bm25_ms"] = t_bm * 1000,
    тогда как в глобалах живут секунды). Из-за этого числа прозы не находились в дампе.
    """
    v = _plain(v)
    if _num(v):
        return v
    if depth >= 4:                     # защита от самоссылающихся структур
        return None
    if isinstance(v, dict) and 0 < len(v) <= 64:
        out = {}
        for _kk, _vv in v.items():
            got = _tree(_vv, depth + 1)
            if got is not None:
                out[str(_kk)] = got
        return out or None
    if isinstance(v, (list, tuple)) and 0 < len(v) <= 64:
        out = [_tree(_x, depth + 1) for _x in v]
        out = [_x for _x in out if _x is not None]
        return out or None
    return None

_metrics = {}
for _k, _v in sorted(globals().items()):
    if _k.startswith("_") or (len(_k) == 1 and _k.islower()):
        continue                       # служебные имена и счётчики циклов
    got = _tree(_v)
    if got is not None:
        _metrics[_k] = got

_payload = {"notebook": NB, "runtime": _runtime, "metrics": _metrics}
_json.dump(_payload, open(RUN_DIR / (NB + ".json"), "w", encoding="utf-8"),
           ensure_ascii=False, indent=1, sort_keys=True)
print(f"величин: {len(_metrics)} · рантайм: {_runtime['gpu'] or 'CPU'}")
_sys.stdout.flush(); _LOG.flush()      # журнал дописан до того, как попадёт в архив

_zip = _shutil.make_archive(str(RUN_DIR), "zip", str(RUN_DIR))
_kb = _os.path.getsize(_zip) / 1024
print(f"архив прогона: {_zip} ({_kb:.0f} КБ)")

# Ссылка на скачивание. Файл живёт в песочнице рантайма и сам до репозитория не доедет,
# а вывод ячейки — доедет: жми ссылку, архив упадёт в загрузки браузера.
_b = _b64.b64encode(open(_zip, "rb").read()).decode()
display(HTML(
    f'<a download="{_os.path.basename(_zip)}" href="data:application/zip;base64,{_b}" '
    f'style="display:inline-block;padding:10px 16px;margin:6px 0;background:#2b4a8b;'
    f'color:#fff;border-radius:4px;text-decoration:none;font-family:sans-serif">'
    f'&#10515; Скачать прогон — {_os.path.basename(_zip)} ({_kb:.0f} КБ)</a>'))
print("скачай архив по ссылке ↑, дальше локально: python3 scripts/import_runs.py")
````
